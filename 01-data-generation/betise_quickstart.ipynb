# BeTiSe Quickstart

Iki fonksiyon: `load_config` ve `generate_dataframe`. Hepsi bu.

```python
import matplotlib.pyplot as plt
from betise import generate_dataframe
from betise.config import load_config
```

## 1. Hicbir parametre vermeden

`load_config()` paket default'unu yukler -> 1 seri uretir.

```python
cfg = load_config()
df, ctx = generate_dataframe(cfg)

plt.figure(figsize=(10, 3))
plt.plot(df['time'], df['data'])
plt.title(ctx['label'])
plt.show()
```

## 2. Baz seri secimi

`dataset` parametresi ile override yap.

```python
cfg = load_config(dataset={'base_series': 'white_noise', 'num_series': 3})
df, ctx = generate_dataframe(cfg)

plt.figure(figsize=(10, 3))
for sid in df['series_id'].unique():
    s = df[df['series_id'] == sid]
    plt.plot(s['time'], s['data'], lw=0.8)
plt.title(ctx['label'])
plt.show()
```

## 3. Trend ekleme

```python
cfg = load_config(dataset={
    'base_series': 'white_noise',
    'num_series': 3,
    'features': {'linear_trend': {'enabled': True, 'direction': 'upward'}},
})
df, ctx = generate_dataframe(cfg)

plt.figure(figsize=(10, 3))
for sid in df['series_id'].unique():
    s = df[df['series_id'] == sid]
    plt.plot(s['time'], s['data'], lw=0.8)
plt.title(ctx['label'])
plt.show()
```
