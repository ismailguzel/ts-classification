# BeTiSe Quickstart

Two functions: `load_config` and `generate_dataframe`. That's all.

```python
import matplotlib.pyplot as plt
from betise import generate_dataframe
from betise.config import load_config
```

## 1. With no parameters

`load_config()` loads the package default -> produces 1 series.

```python
cfg = load_config()
df, ctx = generate_dataframe(cfg)

plt.figure(figsize=(10, 3))
plt.plot(df['time'], df['data'])
plt.title(ctx['label'])
plt.show()
```

## 2. Choosing a base series

Override via the `dataset` parameter.

```python
cfg = load_config(dataset={'base_series': 'white_noise', 'num_series': 3})
df, ctx = generate_dataframe(cfg)

plt.figure(figsize=(10, 3))
for sid in df['series_id'].unique():
    s = df[df['series_id'] == sid]
    plt.plot(s['time'], s['data'], lw=0.8)
plt.title(ctx['label'])
plt.show()
```

## 3. Adding a trend

```python
cfg = load_config(dataset={
    'base_series': 'white_noise',
    'num_series': 3,
    'features': {'linear_trend': {'enabled': True, 'direction': 'upward'}},
})
df, ctx = generate_dataframe(cfg)

plt.figure(figsize=(10, 3))
for sid in df['series_id'].unique():
    s = df[df['series_id'] == sid]
    plt.plot(s['time'], s['data'], lw=0.8)
plt.title(ctx['label'])
plt.show()
```
